```python
import pandas as pd
import numpy as np
import xarray as xr
from scipy.stats import linregress
import matplotlib as mpl
import matplotlib.pyplot as plt
from mpl_toolkits.basemap import Basemap

DATA_DIR = '/home/lachale/Documentos/GitHub/BRISES/data/'
START_DATE = "1981-01-01"
END_DATE = "2021-12-31"

# Coordinates
data_coords = pd.read_csv(DATA_DIR + 'ERA5_Points_MED_landsea.csv',  encoding='latin-1', sep=';')

countries = ['France','Italy','EI','NA','Spain','BI']
country_index = [['77470','76430','76500','76900','161530'],
    ['161680','162420','162450','162890','163100'],
    ['164290','164050','77650','77800','77610','165600','165460','165200'],
    ['604900','603900','604020','603600','607150'],
    ['6155A','6325O','7031','8025','8019','8416Y','8414A','8500A','9981A','0016A','76','0200E'],
    ['B893','B228','B278','B954']]
```

```python
def compute_regional_trend(da, alpha=0.05, decimals=3):
    """Compute linear trend for a regional mean time series."""
    years = pd.to_datetime(da.dates.values).year
    res = linregress(years, da.values)
    slope_decade = res.slope * 10
    signif = res.pvalue < alpha
    trend = "increasing" if signif and res.slope > 0 else "decreasing" if signif and res.slope < 0 else "no trend"
    label = '***' if res.pvalue < 0.001 else '**' if res.pvalue < 0.01 else '*' if res.pvalue < 0.05 else '.' if res.pvalue < 0.1 else 'ns'
    return pd.DataFrame([dict(trend=trend, slope_decade=round(slope_decade,decimals), p_value=round(res.pvalue, decimals), significant=signif, signif_label=label)])

def compute_trends(da, alpha=0.05, decimals=3):
    """Compute linear regression trends per station."""
    out = []
    for st in da.station:
        s = da.sel(station=st).dropna('dates')
        if s.size == 0: continue
        y = pd.to_datetime(s.dates.values).year
        r = linregress(y, s.values)
        tr = "increasing" if r.pvalue<alpha and r.slope>0 else "decreasing" if r.pvalue<alpha and r.slope<0 else "no trend"
        lab = '***' if r.pvalue<.001 else '**' if r.pvalue<.01 else '*' if r.pvalue<.05 else '.' if r.pvalue<.1 else 'ns'
        out.append([st.item(), tr, round(r.slope*10, decimals), round(r.pvalue,decimals), r.pvalue<alpha, lab])
    return pd.DataFrame(out, columns=['station','trend','slope_decade','p_value','significant','signif_label'])


# Summary of stations showing significant trends
def summarize_trend_significance(df, alpha=0.05):
    """Summarize % of stations by trend significance and direction."""
    t = len(df)
    s = lambda c: [c.sum(), round(100*c.sum()/t,1)]
    p = df.p_value <= alpha
    summ = {
        'Significant Positive': s(p & (df.slope_decade > 0)),
        'Significant Negative': s(p & (df.slope_decade < 0)),
        'Non-significant Positive': s(~p & (df.slope_decade > 0)),
        'Non-significant Negative': s(~p & (df.slope_decade < 0))
    }
    return pd.DataFrame(summ, index=['Count','Percentage']).T
```

### **SBs speeds**


```python
# Load SBs speeds monthly aggregates after SBs days filtering on wind speed data
df_ws = pd.read_csv(DATA_DIR + 'SBs_speeds_monthly_aggr.csv', index_col='dates') 
df_ws.index = pd.to_datetime(df_ws.index)

# Convert to xarray Dataset
ds_ws = xr.Dataset(
    {"ws": (("dates","station"), df_ws.values)},
    coords={"dates": df_ws.index, "station": df_ws.columns}
)

# ------------------------------------------------------
# -------------- Regional-scale
# ------------------------------------------------------

# Regional average
da_speeds_annual_regional = ds_ws["ws"].mean(dim="station").resample(dates="Y").mean()
da_speeds_seasonal_regional = ds_ws["ws"].mean(dim="station").resample(dates="Q-NOV").mean()
da_speeds_seasonal_regional = da_speeds_seasonal_regional.drop_sel(dates=np.datetime64("2022-02-28"))

# Climatology
clim_speeds_annual_regional = da_speeds_annual_regional.sel(dates=slice("1981", "2021")).mean("dates")
clim_speeds_seasonal_regional = da_speeds_seasonal_regional.sel(dates=slice("1981", "2021")).groupby("dates.season").mean("dates")

# Anomalies
anom_speeds_annual_regional = da_speeds_annual_regional - clim_speeds_annual_regional
anom_speeds_seasonal_regional = da_speeds_seasonal_regional.groupby("dates.season") - clim_speeds_seasonal_regional

# ------------------------------------------------------
# -------------- Station-scale
# ------------------------------------------------------

# Annual & seasonal means per station
da_speeds_annual = ds_ws["ws"].resample(dates="Y").mean()
da_speeds_seasonal = ds_ws["ws"].resample(dates="Q-NOV").mean()
da_speeds_seasonal = da_speeds_seasonal.drop_sel(dates=np.datetime64("2022-02-28"))

# Annual & seasonal anomalies
clim_speeds_annual = da_speeds_annual.sel(dates=slice("1981", "2021")).mean("dates")
anom_speeds_annual = da_speeds_annual - clim_speeds_annual

clim_speeds_seasonal = da_speeds_seasonal.sel(dates=slice("1981", "2021")).groupby("dates.season").mean("dates")
anom_speeds_seasonal = da_speeds_seasonal.groupby("dates.season") - clim_speeds_seasonal

# ------------------------------------------------------
# -------------- Coastal-scale 
# ------------------------------------------------------

coast_groups = dict(zip(countries, country_index))

# Station averages per coast
ds_coasts = xr.concat(
    [ds_ws["ws"].sel(station=[s for s in st if s in ds_ws.station]).mean("station")
     for st in coast_groups.values()], dim="station").assign_coords(station=list(coast_groups.keys()))

# Annual & seasonal means
da_speeds_annual_coastal = ds_coasts.resample(dates="Y").mean()
da_speeds_seasonal_coastal = ds_coasts.resample(dates="Q-NOV").mean().drop_sel(dates="2022-02-28")

# Climatologies (1981–2021)
clim_speeds_annual_coastal = da_speeds_annual_coastal.sel(dates=slice("1981", "2021")).mean("dates")
clim_speeds_seasonal_coastal = (da_speeds_seasonal_coastal.sel(dates=slice("1981", "2021")).groupby("dates.season").mean("dates"))

# Anomalies
anom_speeds_annual_coastal = da_speeds_annual_coastal - clim_speeds_annual_coastal
anom_speeds_seasonal_coastal = (da_speeds_seasonal_coastal.groupby("dates.season") - clim_speeds_seasonal_coastal)
```

```python
# Trend assessment

# ------------------------------------------------------
# -------------- Regional-scale
# ------------------------------------------------------

# Annual trends
annual_regional_trends_speeds = compute_regional_trend(anom_speeds_annual_regional, decimals=3)

# Seasonal trends
seasons = ['DJF', 'MAM','JJA','SON']
seasonal_regional_trends_speeds = []

for season, da_season in anom_speeds_seasonal_regional.groupby('season'):
    df = compute_regional_trend(da_season, decimals=3)
    df['season'] = season
    seasonal_regional_trends_speeds.append(df)
    
seasonal_regional_trends_speeds = pd.concat(seasonal_regional_trends_speeds)


# ------------------------------------------------------
# -------------- Station-scale
# ------------------------------------------------------

# Annual trends
annual_spatial_trends_speeds = compute_trends(anom_speeds_annual, decimals=3)
# significance
summary_annual_spatial_trends_speeds = summarize_trend_significance(annual_spatial_trends_speeds)

# Seasonal trends
seasonal_spatial_trends_speeds = []
summary_seasonal_spatial_trends_speeds = []

for season, da_season in anom_speeds_seasonal.groupby('season'):
    df = compute_trends(da_season, decimals=3)
    df['season'] = season
    seasonal_spatial_trends_speeds.append(df)
    sum_df = summarize_trend_significance(df)
    summary_seasonal_spatial_trends_speeds.append(sum_df)
    
seasonal_spatial_trends_speeds = pd.concat(seasonal_spatial_trends_speeds, ignore_index=True)
summary_seasonal_spatial_trends_speeds = pd.concat(summary_seasonal_spatial_trends_speeds, keys=seasons)

# ------------------------------------------------------
# -------------- Coastal-scale 
# ------------------------------------------------------
# Annual trends
annual_coastal_trends_speeds = compute_trends(anom_speeds_annual_coastal, decimals=3)

# Seasonal trends
seasonal_coastal_trends_speeds = []

for season, da in anom_speeds_seasonal_coastal.groupby("season"):
    df = compute_trends(da, decimals=3); df["season"] = season
    seasonal_coastal_trends_speeds.append(df)

seasonal_coastal_trends_speeds = pd.concat(seasonal_coastal_trends_speeds, ignore_index=True)
```

```python
def calculate_percentage_change(trend_per_decade, start_year, end_year, reference_value):
    """
    Calculates the total percentage change of a variable over a period.
    """
    # Calculate total decades elapsed
    decades = (end_year - start_year) / 10.0
    
    # Calculate absolute change over the entire period
    absolute_change = trend_per_decade * decades
    
    # Calculate percentage change
    percentage_change = (absolute_change / reference_value) * 100.0
    
    return percentage_change

# Example usage with your data (assuming a reference mean wind speed of 5.0 m/s):
# trend = -0.095
result = calculate_percentage_change(annual_regional_trends_speeds['slope_decade'][0], 1981, 2021, clim_speeds_annual_regional.data)
print(f"Total percentage change: {result}%")

result = calculate_percentage_change(annual_regional_trends_speeds['slope_decade'][0], 1981, 2021, da_speeds_annual_regional[0].data)
print(f"Total percentage change: {result}%")
```

```python
import pandas as pd
import io

def calculate_station_changes(trends_df, clim_df, decades=4.0):
    """
    Merges trend and climatology dataframes to calculate absolute and percentage changes.
    """
    # If 'station' is the index in clim_df, move it to a column
    if 'station' not in clim_df.columns:
        clim_df = clim_df.reset_index(names='station')
        
    # Ensure 'station' columns have the same data type (string) to avoid merge errors
    trends_df['station'] = trends_df['station'].astype(str)
    clim_df['station'] = clim_df['station'].astype(str)
    
    # Merge dataframes on 'station' (inner join keeps only matching stations)
    merged_df = pd.merge(trends_df, clim_df, on='station', how='inner')
    
    # Calculate absolute change over the given decades
    # Note: Using 'slope_decade' based on your provided column name
    merged_df['absolute_change'] = merged_df['slope_decade'] * decades
    
    # Calculate percentage change using climatology ('ws')
    merged_df['percentage_change'] = (merged_df['absolute_change'] / merged_df['ws']) * 100.0
    
    # Filter and order the final output columns
    final_df = merged_df[['station', 'absolute_change', 'percentage_change']].copy()
    
    return final_df

# Execute function
result_dataframe = calculate_station_changes(annual_spatial_trends_speeds, clim_speeds_annual.to_dataframe().reset_index())
# Display result
print(result_dataframe)
```

### Plots _a_ and _b_

```python
def plot_trend_maps(df_speeds_annual, df_speeds_seasonal,
                    df_days_annual, df_days_seasonal,
                    coords, summary_speeds, summary_days, alpha=0.05):
    
    from matplotlib.colors import to_rgba

    lat, lon = coords['Lat'].values, coords['Lon'].values
    seasons, codes = ['Annual','Winter (DJF)','Spring (MAM)','Summer (JJA)','Autumn (SON)'],['Annual','DJF','MAM','JJA','SON']
    fig, axes = plt.subplots(5,2,figsize=(9,13))
    plt.subplots_adjust(wspace=0.04, hspace=0.03)

    colors = ['#ff0000','#0000ff','white','white']
    edge_colors = ['#ff0000','#0000ff','#e00a94','#00aaff']

    # Bins for sizes
    speed_bins = np.arange(-0.6,0.31,0.1)
    occ_bins = np.arange(-3.5,3.1,0.5)
    min_size,max_size =15,60

    def assign_size(df,col,bins):
        df['category'] = pd.cut(df[col], bins=bins, labels=np.arange(len(bins)-1))
        df['size'] = df['category'].astype(float).apply(lambda x: min_size + x*(max_size-min_size)/(len(bins)-2))
        return df

    def plot_map(ax, df, colors, edge_colors, bins, title=None, is_speed=True):
        if df is None or len(df)==0: return
        m = Basemap(projection='merc',llcrnrlon=-10.7,urcrnrlon=18.8,llcrnrlat=34,urcrnrlat=46,resolution='i',ax=ax)
        m.fillcontinents(color='#DCDCDC',alpha=0.6,zorder=3)
        ax.set_title(title,fontsize=10,pad=4); ax.set_xticks([]); ax.set_yticks([])
        [s.set_visible(False) for s in ax.spines.values()]; ax.set_facecolor('white')

        if 'Lat' not in df.columns: df['Lat'],df['Lon']=lat[:len(df)],lon[:len(df)]
        df['sig'],df['pos']=df['p_value']<=alpha, df['slope_decade']>=+0
        df=assign_size(df,'slope_decade',bins)

        def scat(sub,marker,fc,ec,a=0.9):
            if len(sub)==0: return
            x,y = m(sub['Lon'],sub['Lat'])
            m.scatter(x,y,s=sub['size'],marker=marker,facecolors=fc,edgecolors=ec,linewidths=0.8,zorder=5,alpha=a)

        scat(df[df.pos & df.sig], '^', colors[0], edge_colors[0], a=0.8)
        scat(df[df.pos & ~df.sig], '^', colors[2], edge_colors[2], a=0.8)
        scat(df[~df.pos & df.sig], 'v', colors[1], edge_colors[1],a=0.8)
        scat(df[~df.pos & ~df.sig], 'v', colors[3], edge_colors[3], a=0.8)

    def plot_pie(ax, summary_df):
        vals = [summary_df.loc['Significant Positive','Percentage'],
                summary_df.loc['Non-significant Positive','Percentage'],
                summary_df.loc['Significant Negative','Percentage'],
                summary_df.loc['Non-significant Negative','Percentage']]
        colors_pie = [to_rgba('#ff0000',1),to_rgba('#e00a94',0.7),to_rgba('#0000ff',1),to_rgba('#00aaff',0.7)]
        wedges,_ = ax.pie(vals,colors=colors_pie,startangle=90,radius=0.85,wedgeprops=dict(width=0.4,edgecolor='white',linewidth=1))
        ax.set_aspect('equal'); ax.set_xlim(-1,1.7); ax.axis('off')

    # --- Panels
    for i,(label,code) in enumerate(zip(seasons,codes)):
        if code=='Annual':
            df_s,df_d = df_speeds_annual, df_days_annual
            sum_s,sum_d = summary_speeds['Annual'], summary_days['Annual']
        else:
            df_s=df_speeds_seasonal[df_speeds_seasonal['season']==code]
            df_d=df_days_seasonal[df_days_seasonal['season']==code]
            sum_s,sum_d=summary_speeds[code],summary_days[code]

        plot_map(axes[i,0],df_s,colors,edge_colors,speed_bins,is_speed=True)
        plot_map(axes[i,1],df_d,colors,edge_colors,occ_bins,is_speed=False)
        inset1=fig.add_axes([axes[i,0].get_position().x1-0.05,axes[i,0].get_position().y0+0.09,0.06,0.06])
        inset2=fig.add_axes([axes[i,1].get_position().x1-0.05,axes[i,1].get_position().y0+0.09,0.06,0.06])
        plot_pie(inset1,sum_s)
        plot_pie(inset2,sum_d)

    axes[0,0].text(0.03,1.05,"(a) SBs speeds",transform=axes[0,0].transAxes,fontsize=11,fontweight='bold',ha='left',va='bottom')
    axes[0,1].text(0.03,1.05,"(b) SBs occurrence",transform=axes[0,1].transAxes,fontsize=11,fontweight='bold',ha='left',va='bottom')

    for i,label in enumerate(seasons):
        axes[i,0].text(-0.12,0.92,label,transform=axes[i,0].transAxes,fontsize=10,fontweight='bold',ha='center',va='top',rotation=90)
    
    plt.show()

    # Save figure
    # fig.savefig(DIR +"Fig_1_spatial_trends.svg",format='svg', dpi=350, bbox_inches='tight')
```

```python
# Plot trend figures 

summary_speeds = {
    'Annual': summary_annual_spatial_trends_speeds,
    'DJF': summary_seasonal_spatial_trends_speeds.loc['DJF'],
    'MAM': summary_seasonal_spatial_trends_speeds.loc['MAM'],
    'JJA': summary_seasonal_spatial_trends_speeds.loc['JJA'],
    'SON': summary_seasonal_spatial_trends_speeds.loc['SON']
}

summary_days = {
    'Annual': summary_annual_spatial_trends_days,
    'DJF': summary_seasonal_spatial_trends_days.loc['DJF'],
    'MAM': summary_seasonal_spatial_trends_days.loc['MAM'],
    'JJA': summary_seasonal_spatial_trends_days.loc['JJA'],
    'SON': summary_seasonal_spatial_trends_days.loc['SON']
}

plot_trend_maps(    
    df_speeds_annual=annual_spatial_trends_speeds,
    df_speeds_seasonal=seasonal_spatial_trends_speeds,
    df_days_annual=annual_spatial_trends_days,
    df_days_seasonal=seasonal_spatial_trends_days,
    coords=data_coords, 
    summary_speeds=summary_speeds, 
    summary_days=summary_days)
```

```python
plt.rc("axes",linewidth=1)

# --- Setup ---
coasts = list(coast_groups.keys())
colors = ['#0CFF03', '#535FC9','#FFE600', '#FF0C0C','#FA15F5','#00FFE6']
x = len(anom_days_annual_coastal.dates)
axy_seasons = ['Annual', 'DJF', 'MAM', 'JJA', 'SON']

fig, axs = plt.subplots(5, 2, figsize=(10, 15), sharex=True, sharey=False)
fig.subplots_adjust(hspace=0.05, wspace=0.15)


# -----------------------------
# Left column: SBs Speeds anomalies
# -----------------------------
for i, season in enumerate(axy_seasons):
    ax = axs[i, 0]
    ax.set_xlim([1979.5, 2022.5])
    ax.set_ylim([-1, 1.5])
    ax.set_yticks([-0.5,0,0.5,1])
    plt.setp(ax.spines.values(), linewidth=1.5)
    ax.tick_params(pad=6, length=6,width=1.5,labelsize=14, direction='in')
    ax.set_xticks([1981.5,1991,2001,2011,2020.5])
    ax.set_xticklabels(['1981','1991','2001','2011','2021'])

    # Coastal anomalies
    
    if season == 'Annual':
        for j, coast in enumerate(coasts):
            y = anom_speeds_annual_coastal.sel(station=coast).values
            ax.plot(np.arange(1981, 1981 + x), y, color=colors[j], alpha=1, lw=1)

    else:
        for j, coast in enumerate(coasts):
            y = anom_speeds_seasonal_coastal.where(anom_speeds_seasonal_coastal['season'] == season, drop=True).sel(station=coast).values
    
            ax.plot(np.arange(1981, 1981 + x), y, color=colors[j], alpha=1, lw=1)

    # Regional anomalies
    
    if season == 'Annual':
        reg = anom_speeds_annual_regional.values
        trend_val = float(annual_regional_trends_speeds['slope_decade'].values[0])
        ax.plot(np.arange(1981, 1981 + x), reg, color='k', lw=1.5, zorder=3)
        ax.text(1980.5, 1.25, "Trend: "+f"{trend_val:+.3f}"+" m/s per decade", fontsize=10, color='black')

    else:
        reg = anom_speeds_seasonal_regional.where(anom_speeds_seasonal_regional['season'] == season, drop=True).values
        trend_val = float(seasonal_regional_trends_speeds.loc[
            seasonal_regional_trends_speeds['season'] == season, 'slope_decade'].values[0])
        ax.plot(np.arange(1981, 1981 + x), reg, color='k', lw=1.5, zorder=3)
        ax.text(1980.5, 1.25, "Trend: "+f"{trend_val:+.3f}"+" m/s per decade", fontsize=10,  color='black')

# -----------------------------
# Right column: SBs Occurrence anomalies
# -----------------------------
for i, season in enumerate(axy_seasons):
    ax = axs[i, 1]
    ax.set_xlim([1979.5, 2022.2])
    ax.set_ylim([-40, 40])
    ax.set_yticks([-30, -15, 0, 15, 30])
    ax.tick_params(pad=6, length=6,width=1.5,labelsize=14, direction='in')
    plt.setp(ax.spines.values(), linewidth=1.5)
    ax.set_xticks([1981.5,1991,2001,2011,2020.5])
    ax.set_xticklabels(['1981','1991','2001','2011','2021'])
    
    # Coastal anomalies
    
    if season == 'Annual':
        for j, coast in enumerate(coasts):
            y = anom_days_annual_coastal.sel(station=coast).values
            ax.plot(np.arange(1981, 1981 + x), y, color=colors[j], alpha=1, lw=1)

    else:
        for j, coast in enumerate(coasts):
            y = anom_days_seasonal_coastal.where(anom_days_seasonal_coastal['season'] == season, drop=True).sel(station=coast).values
    
            ax.plot(np.arange(1981, 1981 + x), y, color=colors[j], alpha=1, lw=1)

    # Regional anomalies
    
    if season == 'Annual':
        reg = anom_days_annual_regional.values
        trend_val = float(annual_regional_trends_days['slope_decade'].values[0])
        ax.plot(np.arange(1981, 1981 + x), reg, color='k', lw=1.5, zorder=3)
        ax.text(1980.5, 32, "Trend: "+f"{trend_val:+.2f}"+" days per decade", fontsize=10, color='black')

    else:
        reg = anom_days_seasonal_regional.where(anom_days_seasonal_regional['season'] == season, drop=True).values
        trend_val = float(seasonal_regional_trends_days.loc[
            seasonal_regional_trends_days['season'] == season, 'slope_decade'].values[0])
        ax.plot(np.arange(1981, 1981 + x), reg, color='k', lw=1.5, zorder=3)
        ax.text(1980.5, 32, "Trend: "+f"{trend_val:+.2f}"+" days per decade", fontsize=10, color='black')

        

axs[0,0].text(0.03,1.05,"(a) SBs speeds",transform=axs[0,0].transAxes,fontsize=12,fontweight='bold',ha='left',va='bottom')
axs[0,1].text(0.03,1.05,"(b) SBs occurrence",transform=axs[0,1].transAxes,fontsize=12,fontweight='bold',ha='left',va='bottom')

for i,label in enumerate(['Annual','Winter (DJF)','Spring (MAM)','Summer (JJA)','Autumn (SON)']):
    axs[i,0].text(-0.16,0.92,label,transform=axs[i,0].transAxes,fontsize=12,fontweight='bold',ha='center',va='top',rotation=90)
    
plt.show()
```
